```python
def classify_preference(preference, question, options):
    option_text = "\n".join(
        f"{chr(65+i)}. {option}"
        for i, option in enumerate(options)
    )

    messages = [
        {
            "role": "system",
            "content": (
                "You are evaluating whether an answer follows a user's preference. "
                "Choose the option that best follows the preference. "
                "Return only one letter: A, B, C, or D."
            )
        },
        {
            "role": "user",
            "content": f"""
User preference:
{preference}

Question:
{question}

Options:
{option_text}

Which option best follows the user's preference?
Return only A, B, C, or D.
"""
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=5,
            do_sample=False
        )

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()

    return response


prediction = classify_preference(
    preference,
    question,
    options
)

print("MODEL PREDICTION:", prediction)
```

```python
import re
import torch

def generate_candidate_options(preference, question):

    messages = [
        {
            "role": "system",
            "content": (
                "You are creating answer choices for a preference-following benchmark. "
                "Generate exactly four realistic and meaningfully different answer choices "
                "for the question. Some choices may satisfy the user's preference and some "
                "may violate it. Do not force exactly one or exactly two correct choices. "
                "Return ONLY four choices, one per line. "
                "Do not use letters, numbers, JSON, explanations, or labels."
            )
        },
        {
            "role": "user",
            "content": f"""
User preference:
{preference}

Question:
{question}

Generate exactly four realistic answer choices.
"""
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    print("Generating options...")

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=200,
            do_sample=False
        )

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()

    print("\nRAW OUTPUT:")
    print(response)

    # Convert output into individual options
    options = []

    for line in response.splitlines():
        line = line.strip()

        if not line:
            continue

        # Remove accidental A., B., 1., 2., etc.
        line = re.sub(r'^[A-Da-d][.)]\s*', '', line)
        line = re.sub(r'^\d+[.)]\s*', '', line)

        if line:
            options.append(line)

    if len(options) < 4:
        raise ValueError(
            f"Only {len(options)} options generated. "
            f"Raw output:\n{response}"
        )

    return options[:4]


candidate_options = generate_candidate_options(
    naturalized,
    question
)

print("\nFINAL CANDIDATE OPTIONS:")
for i, option in enumerate(candidate_options):
    print(f"{chr(65+i)}. {option}")
```

```python
def identify_valid_options(preference, question, options):

    option_text = "\n".join(
        f"{chr(65+i)}. {option}"
        for i, option in enumerate(options)
    )

    messages = [
        {
            "role": "system",
            "content": (
                "You are a strict preference evaluator. "
                "For each answer option, determine whether it satisfies "
                "the user's preference. "
                "Return ONLY the letters of all compatible options, "
                "separated by commas. "
                "Do not include explanations."
            )
        },
        {
            "role": "user",
            "content": f"""
User preference:
{preference}

Question:
{question}

Options:
{option_text}

Which options are compatible with the user's preference?

Return all compatible option letters.
Example:
A,C

If only one is compatible:
B

If all are compatible:
A,B,C,D
"""
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=20,
            do_sample=False
        )

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()

    print("VALID-OPTION MODEL OUTPUT:")
    print(response)

    letters = re.findall(r'[ABCD]', response.upper())

    # Remove duplicates while preserving order
    letters = list(dict.fromkeys(letters))

    valid_indices = [
        ord(letter) - ord('A')
        for letter in letters
    ]

    return valid_indices


valid_indices = identify_valid_options(
    naturalized,
    question,
    candidate_options
)

print("\nVALID INDICES:", valid_indices)

print("\nVALID OPTIONS:")
for i in valid_indices:
    print(f"{chr(65+i)}. {candidate_options[i]}")
```

```python
import json
import os
import re
import torch
import random

BASE = "/kaggle/working/PrefEval-main/benchmark_dataset"

mcq_path = os.path.join(
    BASE,
    "mcq_options",
    "shop_motors.json"
)

with open(mcq_path, "r", encoding="utf-8") as f:
    original_data = json.load(f)


def generate_naturalized(preference):

    messages = [
        {
            "role": "system",
            "content": (
                "Rewrite the user's preference in natural conversational language. "
                "Preserve the exact meaning and every constraint. "
                "Do not weaken, strengthen, add, or remove requirements. "
                "Return only the rewritten preference."
            )
        },
        {
            "role": "user",
            "content": f"""
Original preference:
{preference}

Rewrite it naturally.
"""
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=100,
            do_sample=False
        )

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    return tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()


def generate_options(question):

    messages = [
        {
            "role": "system",
            "content": (
                "Generate exactly four realistic and meaningfully different "
                "answer choices for the question. "
                "Do not consider any user's personal preference. "
                "The choices should represent different plausible answers. "
                "Return ONLY four choices, one per line. "
                "Do not use letters, numbers, labels, explanations, or JSON."
            )
        },
        {
            "role": "user",
            "content": f"""
Question:
{question}

Generate four realistic answer choices.
"""
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=200,
            do_sample=False
        )

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()

    options = []

    for line in response.splitlines():

        line = line.strip()

        if not line:
            continue

        line = re.sub(r'^[A-Da-d][.)]\s*', '', line)
        line = re.sub(r'^\d+[.)]\s*', '', line)

        if line:
            options.append(line)

    if len(options) < 4:
        raise ValueError(
            f"Could not generate 4 options.\n{response}"
        )

    return options[:4]


def identify_valid_options(preference, question, options):

    option_text = "\n".join(
        f"{chr(65+i)}. {option}"
        for i, option in enumerate(options)
    )

    messages = [
        {
            "role": "system",
            "content": (
                "Evaluate which answer choices are compatible with the user's "
                "preference. There may be one, multiple, or all valid choices. "
                "Do not force a particular number of valid choices. "
                "Return ONLY the letters of all compatible options separated by commas. "
                "No explanation."
            )
        },
        {
            "role": "user",
            "content": f"""
User preference:
{preference}

Question:
{question}

Options:
{option_text}

Return all compatible option letters.
"""
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=20,
            do_sample=False
        )

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()

    letters = re.findall(r'[ABCD]', response.upper())
    letters = list(dict.fromkeys(letters))

    valid_indices = [
        ord(letter) - ord("A")
        for letter in letters
    ]

    return valid_indices


# --------------------------------------------------
# Generate pilot
# --------------------------------------------------

pilot = []

for idx, item in enumerate(original_data[:10]):

    print(f"\n{'='*60}")
    print(f"EXAMPLE {idx + 1}/10")

    preference = item["preference"]
    question = item["question"]

    print("Generating naturalized preference...")
    naturalized = generate_naturalized(preference)

    print("Generating candidate options...")
    options = generate_options(question)

    print("Identifying compatible options...")
    valid_indices = identify_valid_options(
        naturalized,
        question,
        options
    )

    record = {
        "original_id": item.get("id"),
        "preference_original": preference,
        "preference_naturalized": naturalized,
        "question": question,
        "options": options,
        "valid_indices": valid_indices
    }

    pilot.append(record)

    print("Naturalized:", naturalized)

    for i, option in enumerate(options):
        mark = "✓" if i in valid_indices else "✗"
        print(f"{chr(65+i)}. {option} {mark}")

    print("Valid indices:", valid_indices)


# Save pilot
pilot_path = "/kaggle/working/enhanced_shop_motors_pilot.json"

with open(pilot_path, "w", encoding="utf-8") as f:
    json.dump(pilot, f, indent=2, ensure_ascii=False)

print("\n" + "="*60)
print("PILOT COMPLETE")
print("Saved to:", pilot_path)
print("Examples:", len(pilot))
```

```python
def validate_options(preference, question, options):

    results = []

    for i, option in enumerate(options):

        messages = [
            {
                "role": "system",
                "content": (
                    "You are a strict preference evaluator. "
                    "Determine whether the specific answer option satisfies "
                    "the user's preference. "
                    "Consider explicit constraints carefully. "
                    "If the preference excludes a category, brand, feature, "
                    "or property, an option belonging to that excluded category "
                    "must be marked NO. "
                    "Return only YES or NO."
                )
            },
            {
                "role": "user",
                "content": f"""
User preference:
{preference}

Question:
{question}

Answer option:
{option}

Does this option satisfy the user's preference?

Return only:
YES
or
NO
"""
            }
        ]

        text = tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )

        inputs = tokenizer(
            text,
            return_tensors="pt"
        ).to(model.device)

        with torch.inference_mode():
            outputs = model.generate(
                **inputs,
                max_new_tokens=5,
                do_sample=False
            )

        generated = outputs[0][inputs["input_ids"].shape[1]:]

        response = tokenizer.decode(
            generated,
            skip_special_tokens=True
        ).strip().upper()

        if "YES" in response and "NO" not in response:
            valid = True
        elif "NO" in response:
            valid = False
        else:
            valid = None

        results.append({
            "option": option,
            "response": response,
            "valid": valid
        })

    valid_indices = [
        i for i, result in enumerate(results)
        if result["valid"] is True
    ]

    return results, valid_indices
```

```python
import re

def generate_question_options(question):
    messages = [
        {
            "role": "system",
            "content": (
                "You generate answer choices for a multiple-choice question.\n"
                "Generate exactly FOUR realistic and meaningfully different answers "
                "that directly answer the question.\n\n"
                "Important:\n"
                "- Do NOT consider any user's personal preferences.\n"
                "- Do NOT try to make an option satisfy or violate a preference.\n"
                "- Generate options based ONLY on the question.\n"
                "- Options must be realistic recommendations for the situation in the question.\n"
                "- If the question asks about vehicles, give actual vehicles appropriate "
                "to the question, not random sports cars or unrelated models.\n"
                "- Avoid four nearly identical options.\n"
                "- Do not include explanations.\n\n"
                "Return ONLY four options, one per line."
            )
        },
        {
            "role": "user",
            "content": f"Question:\n{question}\n\nGenerate four realistic answer choices."
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=150,
            do_sample=True,
            temperature=0.8,
            top_p=0.9
        )

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()

    print("RAW OUTPUT:")
    print(response)

    options = []

    for line in response.splitlines():
        line = line.strip()

        if not line:
            continue

        # Remove A. / B. / 1. etc.
        line = re.sub(r'^[A-Da-d][\.\):\-]\s*', '', line)
        line = re.sub(r'^\d+[\.\):\-]\s*', '', line)

        if line:
            options.append(line)

    # Remove duplicates
    unique_options = []
    seen = set()

    for option in options:
        key = option.lower().strip()

        if key not in seen:
            seen.add(key)
            unique_options.append(option)

    if len(unique_options) < 4:
        raise ValueError(
            f"Model generated fewer than 4 options:\n{response}"
        )

    return unique_options[:4]
```

```python
def generate_additional_vehicle_candidates(question):
    messages = [
        {
            "role": "system",
            "content": (
                "Generate exactly EIGHT additional real vehicle models "
                "that could reasonably answer the question.\n\n"
                "Do not repeat common choices such as Toyota Corolla, "
                "Honda Civic, Honda Accord, or Tesla Model 3.\n"
                "Use a wide range of vehicle categories and sizes.\n"
                "Include SUVs, three-row SUVs, minivans, crossovers, "
                "pickup trucks, hybrids, and EVs where appropriate.\n"
                "Use real production vehicles.\n"
                "Do not consider any user's personal preference.\n"
                "Return ONLY vehicle names, one per line."
            )
        },
        {
            "role": "user",
            "content": question
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=180,
            do_sample=True,
            temperature=1.2,
            top_p=0.95
        )

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()

    print("ADDITIONAL RAW OUTPUT:")
    print(response)

    candidates = []

    for line in response.splitlines():
        line = line.strip()

        if not line:
            continue

        line = re.sub(
            r'^[A-Ha-h][\.\):\-]\s*',
            '',
            line
        )

        line = re.sub(
            r'^\d+[\.\):\-]\s*',
            '',
            line
        )

        if line:
            candidates.append(line)

    # Remove duplicates
    unique = []
    seen = set()

    for x in candidates:
        key = x.lower().strip()

        if key not in seen:
            seen.add(key)
            unique.append(x)

    if len(unique) < 8:
        raise ValueError(
            f"Only generated {len(unique)} candidates."
        )

    return unique[:8]
```

```python
def build_enhanced_example(example, max_attempts=3):

    preference = example["preference"]
    question = example["question"]

    original_options = example["classification_task_options"]

    # In the original PREFEVAL MCQ construction,
    # option 0 is the preference-compatible option.
    original_valid = [original_options[0]]
    original_invalid = original_options[1:]

    for attempt in range(1, max_attempts + 1):

        print(f"  Attempt {attempt}/{max_attempts}")

        try:

            # Generate additional candidates WITHOUT preference
            candidates = generate_candidate_pool(
                question,
                num_candidates=12
            )

            # Remove exact duplicates of original options
            original_normalized = {
                x.lower().strip()
                for x in original_options
            }

            candidates = [
                x for x in candidates
                if x.lower().strip() not in original_normalized
            ]

            valid_candidates = []
            invalid_candidates = []

            # Validate generated candidates
            for option in candidates:

                is_valid = validate_candidate(
                    preference,
                    question,
                    option
                )

                if is_valid is True:
                    valid_candidates.append(option)

                elif is_valid is False:
                    invalid_candidates.append(option)

            print(
                f"    New valid: {len(valid_candidates)} | "
                f"New invalid: {len(invalid_candidates)}"
            )

            # ------------------------------------------------
            # IMPORTANT:
            # Original compatible option is always retained.
            # New compatible options are added only if the
            # validator independently confirms them.
            # ------------------------------------------------

            all_valid = original_valid + valid_candidates

            # Maximum 4 options in final MCQ
            selected_valid = all_valid[:4]

            needed_invalid = 4 - len(selected_valid)

            # Prefer original high-quality distractors
            selected_invalid = original_invalid[:needed_invalid]

            # If more are needed, use generated invalid options
            if len(selected_invalid) < needed_invalid:

                extra_needed = (
                    needed_invalid - len(selected_invalid)
                )

                selected_invalid += invalid_candidates[:extra_needed]

            # Still not enough?
            if len(selected_invalid) < needed_invalid:

                print(
                    "    Not enough invalid options. Retrying..."
                )

                continue

            # Final four options
            selected = (
                selected_valid +
                selected_invalid[:needed_invalid]
            )

            # Deterministic shuffle
            rng = random.Random(
                1000 + attempt
            )

            rng.shuffle(selected)

            # Determine ground-truth valid indices
            valid_indices = [
                i
                for i, option in enumerate(selected)
                if option in selected_valid
            ]

            return {
                "preference_original": preference,
                "question": question,
                "original_options": original_options,
                "options": selected,
                "valid_indices": valid_indices,
                "num_valid": len(valid_indices),
                "new_valid_candidates": valid_candidates,
                "new_invalid_candidates": invalid_candidates,
                "status": "success",
                "attempt": attempt
            }

        except Exception as e:

            print(f"    ERROR: {e}")

    # Fallback: preserve original benchmark rather than
    # silently inventing ground truth.
    fallback_options = original_options.copy()

    return {
        "preference_original": preference,
        "question": question,
        "original_options": original_options,
        "options": fallback_options,
        "valid_indices": [0],
        "num_valid": 1,
        "new_valid_candidates": [],
        "new_invalid_candidates": [],
        "status": "fallback_original",
        "attempt": max_attempts
    }
```

```python
def verify_naturalization(original, naturalized):

    messages = [
        {
            "role": "system",
            "content": (
                "Compare the ORIGINAL preference and the REWRITTEN preference.\n\n"
                "Check whether the rewritten version preserves ALL of the "
                "original user's constraints, exclusions, priorities, "
                "strength, and conditions.\n\n"
                "Return ONLY:\n"
                "SAME\n"
                "or\n"
                "CHANGED"
            )
        },
        {
            "role": "user",
            "content": f"""
ORIGINAL:
{original}

REWRITTEN:
{naturalized}
"""
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.inference_mode():

        outputs = model.generate(
            **inputs,
            max_new_tokens=5,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    generated = outputs[0][
        inputs["input_ids"].shape[1]:
    ]

    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip().upper()

    if re.search(r"\bSAME\b", response):
        return "SAME"

    if re.search(r"\bCHANGED\b", response):
        return "CHANGED"

    return "UNKNOWN"


# Verify all 45
for item in tqdm(
    naturalized_results,
    desc="Checking semantic preservation"
):

    item["verification"] = verify_naturalization(
        item["original"],
        item["naturalized"]
    )


# Summary
same = sum(
    x["verification"] == "SAME"
    for x in naturalized_results
)

changed = sum(
    x["verification"] == "CHANGED"
    for x in naturalized_results
)

unknown = sum(
    x["verification"] == "UNKNOWN"
    for x in naturalized_results
)

print("\n" + "=" * 60)
print("NATURALIZATION VERIFICATION")
print("=" * 60)

print("SAME:", same)
print("CHANGED:", changed)
print("UNKNOWN:", unknown)

print(
    "Semantic preservation rate:",
    round(100 * same / len(naturalized_results), 2),
    "%"
)
```

```python
import os
import json
import re
import random
from difflib import SequenceMatcher
from tqdm.auto import tqdm

# ============================================================
# SETTINGS
# ============================================================

OUTPUT_PATH = "/kaggle/working/enhanced_mcq_997.json"

MAX_CANDIDATES = 12
MAX_VALID = 4
MAX_OPTIONS = 4
GEN_ATTEMPTS = 3

# ============================================================
# TEXT NORMALIZATION
# ============================================================

def normalize_text(text):
    text = text.lower().strip()
    text = re.sub(r'[^a-z0-9\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    return text


def is_near_duplicate(a, b, threshold=0.88):
    """
    Reject obvious duplicate/paraphrase-like options.
    Uses lexical similarity only, so legitimate different choices
    are not aggressively removed.
    """
    a_norm = normalize_text(a)
    b_norm = normalize_text(b)

    if a_norm == b_norm:
        return True

    ratio = SequenceMatcher(None, a_norm, b_norm).ratio()

    if ratio >= threshold:
        return True

    # Catch cases such as:
    # Tesla Model 3
    # Tesla Model 3 Long Range
    shorter = min(len(a_norm), len(b_norm))
    longer = max(len(a_norm), len(b_norm))

    if shorter > 15 and shorter / longer > 0.65:
        if a_norm in b_norm or b_norm in a_norm:
            return True

    return False


# ============================================================
# CLEAN CANDIDATE TEXT
# ============================================================

def clean_candidate(text):
    text = str(text).strip()

    # Remove markdown
    text = re.sub(r'[*_`]', '', text)

    # Remove option labels
    text = re.sub(
        r'^\s*(?:option\s*)?[A-D][\.\):\-]\s*',
        '',
        text,
        flags=re.IGNORECASE
    )

    # Remove numbering
    text = re.sub(
        r'^\s*\d+[\.\):\-]\s*',
        '',
        text
    )

    # Remove quotes around entire option
    text = text.strip().strip('"').strip("'").strip()

    return text


# ============================================================
# GENERATE QUESTION-ONLY CANDIDATES
# ============================================================

def generate_candidate_pool(question, num_candidates=12):

    prompt = f"""
Generate {num_candidates} diverse, realistic answer choices for the following
user question.

IMPORTANT:
- You are NOT given the user's preference.
- Do NOT assume any hidden preference.
- Generate plausible choices that could naturally answer the question.
- Choices should be meaningfully different from one another.
- Avoid trivial paraphrases or variants of the same item.
- Do not rank the choices.
- Do not mention that preferences are involved.
- Return ONLY a JSON array of strings.

Question:
{question}
"""

    messages = [
        {
            "role": "system",
            "content": "You generate diverse realistic answer choices."
        },
        {
            "role": "user",
            "content": prompt
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=700,
            do_sample=True,
            temperature=1.1,
            top_p=0.95,
            pad_token_id=tokenizer.eos_token_id
        )

    generated = output[0][inputs["input_ids"].shape[1]:]

    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()

    # Extract JSON array
    match = re.search(r'\[[\s\S]*\]', response)

    if not match:
        return []

    try:
        candidates = json.loads(match.group(0))
    except:
        return []

    if not isinstance(candidates, list):
        return []

    cleaned = []

    for candidate in candidates:
        candidate = clean_candidate(candidate)

        if len(candidate) < 5:
            continue

        if candidate not in cleaned:
            cleaned.append(candidate)

    # Remove near duplicates inside generated pool
    final = []

    for candidate in cleaned:
        duplicate = False

        for existing in final:
            if is_near_duplicate(candidate, existing):
                duplicate = True
                break

        if not duplicate:
            final.append(candidate)

    return final[:MAX_CANDIDATES]


# ============================================================
# VALIDATE OPTION AGAINST PREFERENCE
# ============================================================

def validate_candidate(preference, candidate):

    prompt = f"""
Determine whether this answer choice is compatible with the user's preference.

Preference:
{preference}

Answer choice:
{candidate}

Rules:
- YES = the choice satisfies the preference.
- NO = the choice violates or conflicts with the preference.
- Judge only the stated preference.
- Do not infer unrelated preferences.
- Return ONLY YES or NO.
"""

    messages = [
        {
            "role": "system",
            "content": "You are a strict preference compatibility classifier."
        },
        {
            "role": "user",
            "content": prompt
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=5,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    generated = output[0][inputs["input_ids"].shape[1]:]

    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip().upper()

    if re.search(r'\bYES\b', response):
        return True

    if re.search(r'\bNO\b', response):
        return False

    return None


# ============================================================
# BUILD ONE ENHANCED EXAMPLE
# ============================================================

def build_enhanced_example(example):

    preference = example["preference"]
    question = example["question"]

    original_options = example["classification_task_options"]

    # Original benchmark assumes option 0 is the aligned answer
    original_valid = original_options[0]

    for attempt in range(GEN_ATTEMPTS):

        candidates = generate_candidate_pool(
            question,
            MAX_CANDIDATES
        )

        if len(candidates) < 6:
            continue

        # Remove candidates that duplicate original options
        filtered_candidates = []

        for candidate in candidates:

            duplicate = False

            for original in original_options:

                if is_near_duplicate(candidate, original):
                    duplicate = True
                    break

            if not duplicate:
                filtered_candidates.append(candidate)

        # Classify generated candidates
        valid_new = []
        invalid_new = []

        for candidate in filtered_candidates:

            result = validate_candidate(
                preference,
                candidate
            )

            if result is True:
                valid_new.append(candidate)

            elif result is False:
                invalid_new.append(candidate)

        # ----------------------------------------------------
        # Keep original valid answer as guaranteed anchor
        # ----------------------------------------------------

        selected_valid = [original_valid]

        for candidate in valid_new:

            if len(selected_valid) >= MAX_VALID:
                break

            if not any(
                is_near_duplicate(candidate, x)
                for x in selected_valid
            ):
                selected_valid.append(candidate)

        # ----------------------------------------------------
        # Fill remaining slots with invalid choices
        # ----------------------------------------------------

        selected_options = selected_valid.copy()

        # Prefer original invalid distractors first
        original_invalid = original_options[1:]

        for candidate in original_invalid:

            if len(selected_options) >= MAX_OPTIONS:
                break

            if not any(
                is_near_duplicate(candidate, x)
                for x in selected_options
            ):
                selected_options.append(candidate)

        # Then generated invalid choices
        for candidate in invalid_new:

            if len(selected_options) >= MAX_OPTIONS:
                break

            if not any(
                is_near_duplicate(candidate, x)
                for x in selected_options
            ):
                selected_options.append(candidate)

        # Need exactly 4 choices
        if len(selected_options) < MAX_OPTIONS:
            continue

        selected_options = selected_options[:MAX_OPTIONS]

        # Shuffle
        random.shuffle(selected_options)

        valid_indices = [
            i for i, option in enumerate(selected_options)
            if any(
                is_near_duplicate(option, valid)
                for valid in selected_valid
            )
        ]

        # Safety: must contain at least one valid option
        if len(valid_indices) == 0:
            continue

        return {
            "preference": preference,
            "question": question,
            "original_options": original_options,
            "options": selected_options,
            "valid_indices": valid_indices,
            "num_valid": len(valid_indices),
            "new_valid_candidates": valid_new,
            "new_invalid_candidates": invalid_new,
            "status": "success",
            "attempt": attempt + 1
        }

    # --------------------------------------------------------
    # Fallback: original single-valid MCQ
    # --------------------------------------------------------

    options = original_options.copy()
    random.shuffle(options)

    valid_indices = [
        i for i, option in enumerate(options)
        if option == original_valid
    ]

    return {
        "preference": preference,
        "question": question,
        "original_options": original_options,
        "options": options,
        "valid_indices": valid_indices,
        "num_valid": len(valid_indices),
        "new_valid_candidates": [],
        "new_invalid_candidates": [],
        "status": "fallback",
        "attempt": GEN_ATTEMPTS
    }


print("Functions loaded successfully.")
print("Ready to process", len(clean_examples), "clean examples.")
```

```python
import re
import torch
import time

def parse_choices(text):
    """
    Extract A/B/C/D choices from model output.
    Supports:
        A
        A,C
        A, C
        A and C
        The answers are A and C
    """

    text = text.upper()

    # Remove anything inside <choice> tags if present
    text = re.sub(r"<CHOICE>|</CHOICE>", " ", text)

    matches = re.findall(r"\b([ABCD])\b", text)

    # Preserve order and remove duplicates
    choices = []

    for c in matches:
        if c not in choices:
            choices.append(c)

    return choices


def evaluate_mcq(preference, question, options, multi_valid=False):

    if multi_valid:
        instruction = """Select ALL options that are compatible with the user's preference.
There may be one or multiple correct options.

Return ONLY the letters of all compatible options.
Examples:
A
A,C
B,D

Do not provide explanations."""
    else:
        instruction = """Select the ONE option that is most compatible with the user's preference.

Return ONLY one letter:
A
B
C
or
D

Do not provide explanations."""

    option_text = "\n".join(
        f"{chr(65+i)}. {option}"
        for i, option in enumerate(options)
    )

    user_prompt = f"""User preference:
{preference}

Question:
{question}

Options:
{option_text}

{instruction}
"""

    messages = [
        {
            "role": "system",
            "content": "You are evaluating whether answers follow a user's stated preference."
        },
        {
            "role": "user",
            "content": user_prompt
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=8,
            do_sample=False
        )

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()

    predicted_letters = parse_choices(response)

    return {
        "response": response,
        "predicted_letters": predicted_letters
    }


print("Evaluator function ready.")
```

```python
import json
import numpy as np

with open("/kaggle/working/pilot_results_100.json", "r") as f:
    results = json.load(f)

def exact_match(pred, gold):
    return set(pred) == set(gold)

def set_metrics(pred, gold):
    pred_set = set(pred)
    gold_set = set(gold)

    tp = len(pred_set & gold_set)

    precision = tp / len(pred_set) if pred_set else 0.0
    recall = tp / len(gold_set) if gold_set else 0.0

    if precision + recall == 0:
        f1 = 0.0
    else:
        f1 = 2 * precision * recall / (precision + recall)

    return precision, recall, f1


conditions = [
    "C1_original_original",
    "C2_naturalized_original",
    "C3_original_enhanced",
    "C4_naturalized_enhanced"
]

print("=" * 70)
print("PILOT RESULTS — 100 EXAMPLES")
print("=" * 70)

for condition in conditions:
    exact = 0
    precisions = []
    recalls = []
    f1s = []

    for r in results:
        pred = r[condition]["prediction"]
        gold = r[condition]["gold"]

        if exact_match(pred, gold):
            exact += 1

        p, rec, f1 = set_metrics(pred, gold)
        precisions.append(p)
        recalls.append(rec)
        f1s.append(f1)

    print(f"\n{condition}")
    print(f"  Exact Match : {exact}/100 = {exact:.1f}%")
    print(f"  Precision   : {np.mean(precisions)*100:.2f}%")
    print(f"  Recall      : {np.mean(recalls)*100:.2f}%")
    print(f"  F1          : {np.mean(f1s)*100:.2f}%")

print("\n" + "=" * 70)
```

```python
import re

def verify_with_mistral(original, rewritten):

    prompt = f"""Check whether the rewritten user preference preserves the exact
meaning and constraints of the original preference.

The rewrite MAY:
- use synonyms
- sound more casual
- change sentence structure
- shorten the wording

The rewrite MUST NOT:
- add a preference
- remove a preference
- weaken or strengthen a preference
- change exclusions or inclusions
- change quantities, thresholds, or conditions
- contradict the original meaning

ORIGINAL:
{original}

REWRITTEN:
{rewritten}

Answer ONLY one word:
SAME
or
CHANGED
"""

    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ]

    text = verifier_tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = verifier_tokenizer(
        text,
        return_tensors="pt"
    ).to(verifier_model.device)

    with torch.no_grad():
        outputs = verifier_model.generate(
            **inputs,
            max_new_tokens=5,
            do_sample=False
        )

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    result = verifier_tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip().upper()

    if "CHANGED" in result:
        return "CHANGED"
    elif "SAME" in result:
        return "SAME"
    else:
        return "UNKNOWN"
```

```python
import json
import time
import os

results_10turn = []

output_path = "/kaggle/working/eval_10turn_checkpoint.json"

# Resume if checkpoint exists
if os.path.exists(output_path):
    with open(output_path, "r") as f:
        results_10turn = json.load(f)
    print("Resuming from:", len(results_10turn))

start = len(results_10turn)

for i in range(start, len(original_data)):

    mcq = original_data[i]
    nat = final_verified_data[i]

    # Use one of the 24 unrelated conversations
    conversation = inter_turns[i % len(inter_turns)]["conversation"][:20]

    row = {
        "index": i,
        "preference_original": nat["preference_original"],
        "preference_naturalized": nat["preference_final_independent"],
        "question": nat["question"],
        "options": mcq["classification_task_options"],
        "gold": ["A"]   # temporary; corrected below
    }

    # Reproducibly shuffle original options
    import random
    rng = random.Random(42 + i)

    option_pairs = list(enumerate(mcq["classification_task_options"]))
    rng.shuffle(option_pairs)

    shuffled_options = [x[1] for x in option_pairs]
    correct_original_index = 0
    gold_letter = "ABCD"[
        [x[0] for x in option_pairs].index(correct_original_index)
    ]

    row["options"] = shuffled_options
    row["gold"] = [gold_letter]

    for condition, preference in [
        ("original", nat["preference_original"]),
        ("naturalized", nat["preference_final_independent"])
    ]:

        user_prompt = f"""User preference:
{preference}

Question:
{nat["question"]}

Options:
A. {shuffled_options[0]}
B. {shuffled_options[1]}
C. {shuffled_options[2]}
D. {shuffled_options[3]}

Select the ONE option that is most compatible with the user's preference.

Return ONLY one letter:
A
B
C
or
D"""

        messages = [
            {"role": "system", "content": "You are a helpful assistant."}
        ]

        messages.extend(conversation)

        messages.append({
            "role": "user",
            "content": user_prompt
        })

        text = tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )

        inputs = tokenizer(
            text,
            return_tensors="pt"
        ).to(model.device)

        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_new_tokens=5,
                do_sample=False
            )

        generated = outputs[0][inputs["input_ids"].shape[1]:]

        response = tokenizer.decode(
            generated,
            skip_special_tokens=True
        ).strip()

        prediction = parse_choices(response)

        row[condition] = {
            "response": response,
            "prediction": prediction,
            "gold": [gold_letter]
        }

    results_10turn.append(row)

    # Save every 10 examples
    if (i + 1) % 10 == 0:
        with open(output_path, "w") as f:
            json.dump(results_10turn, f, indent=2)

        print(f"[{i+1}/{len(original_data)}] saved")

# Final save
with open(output_path, "w") as f:
    json.dump(results_10turn, f, indent=2)

print("DONE:", len(results_10turn))
```
